```python
import random

def make_blobs(centers, points_per_center, spread, seed=0):
    '''Create a synthetic 2D dataset made of Gaussian blobs.

    Parameters
    ----------
    centers : list of points, e.g. [[0, 0], [5, 5]]
        The true centre of each blob.
    points_per_center : int
        How many points to generate around each centre.
    spread : float
        Standard deviation of the blob (bigger = more spread out).
    seed : int
        Seed for the random number generator, so the result is reproducible.

    Returns
    -------
    points : list of points (the dataset, shuffled)
    true_labels : list of ints (which blob each point came from)
    '''
    random.seed(seed)
    points = []
    true_labels = []

    for blob_index, center in enumerate(centers):
        for _ in range(points_per_center):
            x = random.gauss(center[0], spread)
            y = random.gauss(center[1], spread)
            points.append([x, y])
            true_labels.append(blob_index)

    # Shuffle so that the points are not ordered by blob.
    combined = list(zip(points, true_labels))
    random.shuffle(combined)
    points = [p for p, _ in combined]
    true_labels = [t for _, t in combined]

    return points, true_labels


centers = [[0.0, 0.0], [5.0, 5.0]]
points_per_center = 30
spread = 0.5

points, labels = make_blobs(centers, points_per_center, spread)

print(points)
print(labels)
```
